```python
from google.colab import ai
def personal_Agent():
  while(True):
    prompt=input("Ask any question: ")
    response=ai.generate_text(prompt,model_name="google/gemini-2.5-flash")
    print(response)
    if prompt=="quit":
      break
  ask=input("type 'start' whenever you feel right!")
  if ask=="start"or ask=="Start":
    personal_Agent()
personal_Agent()
```

**Note: The AI doesn't have memory: It don't remembers the past information.**
